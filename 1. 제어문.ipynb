# 반복문 - for

```python
nums = [273, 310, 4, 6, 200, 100, 20, 50, 40, 101]
len(nums)
```

# 100 이상의 숫자만 출력

```python
for num in nums:
    # 100 이상의 숫자만 출력
    if (num >= 100):
        print(f"- 100 이상의 수: {num}")
```

# 100 이상의 숫자, 100 미만인 숫자를 따로 새로운 리스트에 넣어서 각각 리스트로 출력

```python
over = []
under = []

for num in nums:
    if (num >= 100):
        over.append(num)
    else:
        under.append(num)

print(f"- 100 이상의 수: {over}")
print(f"- 100 미만의 수: {under}")
```

## for + dict

```python
# nums의 숫자가 각각 몇번 반복하는지 dict로 만들어서 출력
# {1: n, 2: n ... }

nums = [1,2,6,7,4,3,2,1,5,9,4,9,7,2,1,3,4,5,6,7,5,6,7,5]
counter = {}

for num in nums:
    # 키값 판단
    if num not in counter:
    # 해당 숫자가 제일 처음 나왔을 때, 키값 + 카운트
        counter[num] = 1
    else:
    # 해당 숫자가 처음 나온게 아닐 때 카운트
        counter[num] += 1

print(counter)
```

## 리프트 컴프리헨션(list comprehension)

```python
array = [i * i for i in range(0, 20, 2)]
print(array)
```

```python
# array = [i * i for i in range(0, 20, 2)]
# for 문으로 만들기

array = []
for i in range(0, 20, 2):
    array.append(i * i)

print(array)
```

### 리스트 컴프리헨션 + 조건문

```python
# 짝수만 출력하기
even_nums = []
for i in range(0, 20, 2):
    # i가 짝수인가?
    if(i % 2 == 0):
        even_nums.append(i)

even_nums

```

```python
even_nums = [i for i in range(1, 21) if i % 2 == 0]

even_nums
```

## 제어문 종합문제

```python
# 1~100까지 소수 구하기
# 소수: 1보다 큰 자연수 중 1과 자기자신만을 약수로 가지는 수

decimal = []

for i in range(1, 101):
    if i == 2:
        decimal.append(i)
    deno = i - 1

    for j in range(2, i):
        if i % j == 0:
            break

        if j == deno:
            decimal.append(i)

decimal
```
