vs-code에서 가상환경 세팅을 할 때는 ctrl shift p

```python
import ollama

messages = []

while True:
    user_input = input()

    if user_input.lower() in ["exit", "quit", "종료"]:
        print("채팅을 종료합니다.")
        break

    print("USER:", user_input)
    
    messages.append({
        "role": "user",
        "content": user_input
    })

    response = ollama.chat(
        model="exaone3.5:7.8b",
        messages=messages
    )

    answer = response.message.content

    print("EXAONE:", answer)

    messages.append({
        "role": "assistant",
        "content": answer
    })
```
