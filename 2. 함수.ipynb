```python
# 매개변수로 전달된 값들을 모두 곱해서 리턴하는 가변매개 변수 함수를 만들기
def mul(*values):
    n = 1

    for i in values:
        n = n * i

    return n


print(mul(5, 7, 9, 10))
```

```python
# 두 수를 입력받고 그 수들을 매개변수로 받아 1~100까지의 공배수를 구하는 함수 만들기

def comMul(num1, num2):
    if (num1 == 0 or num2 == 0):
        return 0

    commonMult = []

    for i in range(1, 101):
        #if (i % num1 == 0 and i % num2 == 0):
        if (i % (num1 * num2) == 0):
            commonMult.append(i)

    return commonMult

n1 = int(input("숫자1 입력: "))
n2 = int(input("숫자2 입력: "))
# input_data = input("두 수 입력 ex) 2, 3 > ")
# 이 방법으로 하면 split을 써야함 공백도 고려
# num1, num2 = input_data.split(",")     # int()로 형변환하면 공백이 사라진다?


print(comMul(n1, n2));
```
